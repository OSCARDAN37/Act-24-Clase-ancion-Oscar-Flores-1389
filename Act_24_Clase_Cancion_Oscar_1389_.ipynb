```python
class Cancion:
    def __init__(self, titulo, artista, album, duracion):
        # Atributos de la canción
        self.titulo = titulo
        self.artista = artista
        self.album = album
        self.duracion = duracion
        self.reproducciones = 0  # Atributo extra para control interno

    # --- Funciones (Métodos) ---

    def reproducir(self):
        self.reproducciones += 1
        print(f"\n[▶️] Escuchando: {self.titulo}")
        print(f"De: {self.artista} | Álbum: {self.album} | {self.duracion}")
        print(f"Esta canción se ha reproducido {self.reproducciones} vez/veces.")

    def agregarPlaylist(self):
        nombre_pl = input(f"¿A qué playlist deseas agregar '{self.titulo}'?: ")
        print(f"\n[➕] ¡Listo! Se añadió a la playlist: {nombre_pl}")

    def compartir(self):
        print(f"\n[🔗] Generando enlace de compartir para {self.artista}...")
        print(f"Copiado al portapapeles: music.app/share/{self.titulo.lower().replace(' ', '_')}")

# --- Parte Interactiva (Creación de Objetos) ---

def registrar_cancion(numero):
    print(f"\n--- Configurando Canción {numero} ---")
    titulo = input(f"Título de la canción: ")
    artista = input("Nombre del artista: ")
    album = input("Nombre del álbum: ")
    duracion = input("Duración (ej. 3:45): ")

    return Cancion(titulo, artista, album, duracion)

# Creamos los dos objetos (canciones) mediante inputs
cancion1 = registrar_cancion(1)
cancion2 = registrar_cancion(2)

# Demostración de funcionalidad interactiva
print(f"\n--- Panel de Control: {cancion1.titulo} ---")
cancion1.reproducir()
cancion1.agregarPlaylist()

print(f"\n--- Panel de Control: {cancion2.titulo} ---")
cancion2.reproducir()
cancion2.compartir()

print("\n" + "-"*40)
print("Programa realizado por Oscar Flores NL = 19")
```
